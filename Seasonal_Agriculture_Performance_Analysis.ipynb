Data Cleaning

```python
df_clean = df.copy()
```

```python
missing_cols = [
    'Rainfall_mm',
    'Soil_Moisture_pct',
    'Yield_Tonnes_Ha'
]

for col in missing_cols:
    df_clean[col] = df_clean.groupby('Season')[col].transform(
        lambda x: x.fillna(x.median())
    )

df_clean = df_clean.drop_duplicates()
```

```python
# Create final cleaned dataset

df_clean = df.copy()

print("Cleaned dataset created successfully.")
print("Shape:", df_clean.shape)
print("Missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
```

Fill Missing Rainfall

```python
df_clean['Rainfall_mm'] = df_clean.groupby(
    'Season'
)['Rainfall_mm'].transform(
    lambda x: x.fillna(x.median())
)
```

Fill Missing Soil Moisture

```python
df_clean['Soil_Moisture_pct'] = df_clean.groupby(
    'Season'
)['Soil_Moisture_pct'].transform(
    lambda x: x.fillna(x.median())
)
```

Fill Missing Yield

```python
df_clean['Yield_Tonnes_Ha'] = df_clean.groupby(
    'Season'
)['Yield_Tonnes_Ha'].transform(
    lambda x: x.fillna(x.median())
)
```

Economic Analysis

```python
# Economic Performance by Season

economic_summary = (
    df_clean.groupby('Season')
    .agg({
        'Revenue_INR': 'mean',
        'Total_Cost_INR': 'mean',
        'Profit_INR': 'mean'
    })
    .round(2)
)

print("Economic Performance by Season:")
economic_summary
```



Kharif season shows the highest average revenue and profit among all three seasons. Rabi has moderate revenue and profit, while Zaid shows the lowest average revenue and a negative average profit.

The negative average profit in Zaid indicates that the average production cost is higher than the average revenue during this season. This suggests that Zaid-season farming may require better cost management or improved productivity.

```python
# Revenue, Cost and Profit Comparison by Season

economic_plot = economic_summary.reset_index()

economic_plot.plot(
    x='Season',
    y=['Revenue_INR', 'Total_Cost_INR', 'Profit_INR'],
    kind='bar',
    figsize=(9, 5)
)

plt.title('Revenue, Cost and Profit by Season')
plt.xlabel('Season')
plt.ylabel('Amount (INR)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Profit Margin by Season

profit_margin = (
    df_clean.groupby('Season')
    .apply(lambda x: (x['Profit_INR'].sum() / x['Revenue_INR'].sum()) * 100)
    .round(2)
)

print("Profit Margin by Season:")
print(profit_margin)

plt.figure(figsize=(7, 5))

sns.barplot(
    x=profit_margin.index,
    y=profit_margin.values
)

plt.title('Profit Margin by Season')
plt.xlabel('Season')
plt.ylabel('Profit Margin (%)')
plt.tight_layout()
plt.show()
```

### Economic Analysis – Observation

Kharif season has the highest average revenue and profit, indicating better economic performance compared with Rabi and Zaid.

Rabi shows moderate economic performance with positive average profit.

Zaid has the lowest average revenue and records a negative average profit. This indicates that the average cost of cultivation is higher than the average revenue during the Zaid season.

Overall, the results suggest that seasonal conditions have a noticeable impact on agricultural profitability and that cost management is particularly important for Zaid-season farming.
